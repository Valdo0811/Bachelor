```python
from anomalib.data.dataclasses.torch import ImageBatch, ImageItem
import torch
```

```python



res = torch.load("predictions/metal_nut/flip/flipped_nut.pt", weights_only=False)

pred = res["pred_label"]
gt = res["gt_label"]

correct_label = []
correct_counter = 0
false_counter = 0

for i in range(len(pred)):
    if pred[i] == gt[i]:
        correct_label.append(True)
        correct_counter = correct_counter + 1
    else:
        correct_label.append(False)
        false_counter = false_counter + 1

print(correct_label)
print(correct_counter)
print(false_counter)
print(res["pred_score"])
```
